```python
import pandas as pd
```

```python
df = pd.read_csv("student performance and behaviour.csv")
```

```python
# -------------------------------
# STEP 2: CLEAN COLUMN NAMES
# -------------------------------
df.columns = df.columns.str.strip()


# -------------------------------
# STEP 5: CONVERT ALL LIKERT VALUES
# -------------------------------
scale = {
    "Never": 1,
    "Occasionally": 2,
    "Sometimes": 3,
    "Often": 4,
    "Always": 5
}

for col in df.columns:
    df[col] = df[col].replace(scale)

# -------------------------------
# STEP 6: CREATE STRESS SCORE
# -------------------------------
stress_cols = [
    'Felt Nervous and Stressed?',
    'Unable to control important things in life?',
    'Felt difficulties piling up?',
    'Felt confident handling problems?',
    'Felt things were going your way?'
]

df["stress_score"] = df[stress_cols].sum(axis=1)

# -------------------------------
# STEP 7: CREATE PROCRASTINATION SCORE
# -------------------------------
procrast_cols = [
    'I delay starting my study tasks',
    'I postpone assignments until the last moment',
    'I avoid difficult subjects',
    'I study only before exams',
    'I get distracted while studying',
    'I plan to study but don’t follow the plan'
]

df["procrastination_score"] = df[procrast_cols].sum(axis=1)

# -------------------------------
# STEP 8: SELECT FINAL COLUMNS
# -------------------------------
df = df[['study_hours','sleep_hours','attendance','cgpa','stress_score','procrastination_score']]

# -------------------------------
# STEP 9: REMOVE MISSING VALUES
# -------------------------------
df = df.dropna()

# -------------------------------
# STEP 10: SAVE CLEAN DATA
# -------------------------------
df.to_csv("cleaned_data.csv", index=False)

print("✅ Data cleaned successfully!")
print(df.head())
```

```python
# -------------------------------
# STEP 3: RENAME IMPORTANT COLUMNS
# -------------------------------
df.rename(columns={
    'Graduation Marks': 'cgpa',
    'Graduation Marks ': 'cgpa',
    'Study_Hours': 'study_hours',
    'Sleep_Hours': 'sleep_hours',
    'Attendance': 'attendance',
    'What is your current stress level?': 'stress_level'
}, inplace=True)
```

```python
# -------------------------------
# STEP 4: CONVERT STRESS LEVEL
# -------------------------------
df["stress_level"] = df["stress_level"].replace({
    "Low": 1,
    "Medium": 2,
    "High": 3
})
```
